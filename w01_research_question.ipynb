## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Research Question:** Which high-impression content pieces are suffering from severely low click-through rates (CTR) and represent the highest priority opportunities for a content refresh to recover traffic?

**Decision:** Prioritize specific underperforming articles for immediate optimization rather than creating completely new content from scratch.

**Action:** Rewrite the title tags, update meta descriptions, and refresh the outdated information inside the article to drastically improve search click-through performance.

**Cost of a Wrong Call:** Wasting valuable editorial time and budget rewriting content that has zero search visibility, while leaving massive traffic opportunities like this one to completely decay.


```python
# Section 2 Code: Calculate click-through rate to quantify the problem
if 'df' in globals() and df is not None:
    # Look up the specific underperforming content item we identified
    target_id = "9aa793dad4895"
    row = df[df['content_id'].str.contains(target_id, na=False, case=False)]

    if not row.empty:
        impressions = row['impressions_90d'].values[0]
        clicks = row['clicks_90d'].values[0]
        ctr = (clicks / impressions) * 100

        print(f"Target Content ID: {row['content_id'].values[0][:15]}...")
        print(f"Total Impressions: {impressions:,}")
        print(f"Total Clicks: {clicks}")
        print(f"Calculated CTR: {ctr:.2f}%")
    else:
        print("Target content item not found in the dataframe.")
else:
    print("DataFrame 'df' is not loaded. Please run the first cell up top again!")
```
